```python
import pandas as pd
```

```python
data = pd.read_csv('../data/diamonds.csv')
```

```python
# Duplicated lines
print('Duplicated lines:')
print(data.duplicated().sum())

# Missing values
print('\nMissing values:')
print(data.isnull().sum())

# Carat mean
print('\nCarat mean:')
print(data['carat'].mean())

# Simple sample
print('\nSample of 30 lines:')
print(data.sample(30))

# Samples variables
sample_size = 30
classes = data['color'].unique()
quantity_per_class = round(sample_size / len(classes))
samples_per_class = []

# Uniform stratified sample using diamond color
for c in classes:
 class_index = data['color'] == c

 class_observation = data[class_index]
 class_sample = class_observation.sample(quantity_per_class)

 samples_per_class.append(class_sample)

uniform_stratified_sample = pd.concat(samples_per_class)

print('\nUniform stratified sample:')
print(uniform_stratified_sample)

# Proportional stratified sample using diamond color
for c in classes:
 class_index = data['color'] == c

 class_observation = data[class_index]

 class_proportion = len(class_observation) / len(data)
 class_quantity = round(class_proportion * sample_size)
 class_sample = class_observation.sample(class_quantity)

 samples_per_class.append(class_sample)

proportional_stratified_sample = pd.concat(samples_per_class)

print('\nProportional stratified sample:')
print(proportional_stratified_sample)
```
