```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
import seaborn as sns
```

```python
data = pd.read_excel('../data/fast-food-menu.xlsx')
```

```python
peso = data.peso.apply(lambda x: float(str(x).split(' ')[0]))
fig = plt.figure()
ax1 = fig.add_subplot(211)
prob = stats.probplot(peso, dist=stats.norm, plot=ax1)
ax1.set_xlabel('')
ax1.set_title('Probplot against normal distribution')
plt.show()
```

```python
peso_boxcox, _ = stats.boxcox(peso)
fig = plt.figure()
ax1 = fig.add_subplot(211)
prob = stats.probplot(peso_boxcox, dist=stats.norm, plot=ax1)
ax1.set_xlabel('')
ax1.set_title('Probplot after Box-Cox transformation')
plt.show()
```
